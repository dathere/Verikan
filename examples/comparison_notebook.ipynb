## Setup

```python
import pandas as pd
import altair as alt
import requests
from datetime import datetime

DATA_COMMONS_API = "https://api.datacommons.org"
```

## Step 1: Query Parsing

**Intent:** COMPARISON

**Entities Extracted:**
- Places: California (geoId/06), Texas (geoId/48)
- Variables: GDP (GrossDomesticProduction_RealValue)
- Time: Latest available

```python
entities = {
    "places": [
        {"dcid": "geoId/06", "name": "California", "type": "State"},
        {"dcid": "geoId/48", "name": "Texas", "type": "State"}
    ],
    "variables": [
        {"dcid": "GrossDomesticProduction_RealValue", "name": "GDP", "unit": "USD"}
    ]
}

print("Entities for comparison:")
for place in entities['places']:
    print(f"  - {place['name']} ({place['dcid']})")
```

## Step 2: Data Retrieval

```python
# Retrieved data (demonstration values)
data = pd.DataFrame({
    'State': ['California', 'Texas'],
    'GDP (Trillion USD)': [3.89, 2.36],
    'Year': [2024, 2024],
    'dcid': ['geoId/06', 'geoId/48']
})

print("Retrieved Data:")
display(data)
```

## Step 3: Statistical Computation - Comparison

```python
# Compute comparison statistics
ca_gdp = 3.89
tx_gdp = 2.36

difference = ca_gdp - tx_gdp
pct_difference = ((ca_gdp - tx_gdp) / tx_gdp) * 100

comparison_result = {
    "computation_type": "comparison",
    "highest": {"place": "California", "value": ca_gdp},
    "lowest": {"place": "Texas", "value": tx_gdp},
    "absolute_difference": difference,
    "percentage_difference": pct_difference,
    "ranking": ["California", "Texas"]
}

print(f"Comparison Results:")
print(f"  Highest: {comparison_result['highest']['place']} - ${comparison_result['highest']['value']}T")
print(f"  Lowest: {comparison_result['lowest']['place']} - ${comparison_result['lowest']['value']}T")
print(f"  Difference: ${difference:.2f}T ({pct_difference:.1f}% higher)")
```

## Step 4: Visualization

```python
# Create comparison bar chart
chart = alt.Chart(data).mark_bar().encode(
    x=alt.X('State:N', title='State'),
    y=alt.Y('GDP (Trillion USD):Q', title='GDP (Trillion USD)'),
    color=alt.Color('State:N', legend=None),
    tooltip=['State', 'GDP (Trillion USD)', 'Year']
).properties(
    title='GDP Comparison: California vs Texas (2024)',
    width=400,
    height=300
)

chart
```
