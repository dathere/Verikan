```python
import pandas as pd
import altair as alt
import numpy as np
from datetime import datetime

DATA_COMMONS_API = "https://api.datacommons.org"
```

## Step 1: Query Parsing

**Intent:** TREND_ANALYSIS

**Entities Extracted:**
- Places: Florida (geoId/12)
- Variables: Population (Count_Person)
- Time: 2019-2024 (past 5 years)

```python
entities = {
    "places": [
        {"dcid": "geoId/12", "name": "Florida", "type": "State"}
    ],
    "variables": [
        {"dcid": "Count_Person", "name": "Population", "unit": "persons"}
    ],
    "times": [
        {"start_date": "2019", "end_date": "2024", "period_type": "range"}
    ]
}

print(f"Analyzing trend for: {entities['places'][0]['name']}")
print(f"Variable: {entities['variables'][0]['name']}")
print(f"Period: {entities['times'][0]['start_date']} to {entities['times'][0]['end_date']}")
```

## Step 2: Data Retrieval

```python
# Retrieved time series data (demonstration values)
data = pd.DataFrame({
    'Year': [2019, 2020, 2021, 2022, 2023, 2024],
    'Population (Millions)': [21.48, 21.78, 22.24, 22.61, 22.97, 23.35],
    'State': ['Florida'] * 6
})

print("Retrieved Time Series Data:")
display(data)
```

## Step 3: Statistical Computation - Trend Analysis

```python
# Compute trend statistics
start_value = 21.48
end_value = 23.35

absolute_change = end_value - start_value
percentage_change = ((end_value - start_value) / start_value) * 100

# Calculate year-over-year growth
yoy_growth = data['Population (Millions)'].pct_change() * 100
avg_annual_growth = yoy_growth.mean()

# Determine trend direction
if percentage_change > 5:
    trend_direction = "increasing"
elif percentage_change < -5:
    trend_direction = "decreasing"
else:
    trend_direction = "stable"

trend_result = {
    "computation_type": "trend_analysis",
    "trend_direction": trend_direction,
    "start_value": start_value,
    "end_value": end_value,
    "start_date": "2019",
    "end_date": "2024",
    "absolute_change": absolute_change,
    "percentage_change": percentage_change,
    "average_annual_growth": avg_annual_growth
}

print(f"Trend Analysis Results:")
print(f"  Direction: {trend_direction}")
print(f"  Start ({trend_result['start_date']}): {start_value}M")
print(f"  End ({trend_result['end_date']}): {end_value}M")
print(f"  Total Change: +{absolute_change:.2f}M ({percentage_change:.1f}%)")
print(f"  Avg Annual Growth: {avg_annual_growth:.2f}%")
```

## Step 4: Visualization

```python
# Create trend line chart
line = alt.Chart(data).mark_line(point=True, strokeWidth=3).encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('Population (Millions):Q', 
            title='Population (Millions)',
            scale=alt.Scale(zero=False)),
    tooltip=['Year', 'Population (Millions)']
).properties(
    title='Florida Population Trend (2019-2024)',
    width=500,
    height=300
)

# Add trend annotation
text = alt.Chart(pd.DataFrame({
    'x': [2024],
    'y': [23.35],
    'text': ['+8.7%']
})).mark_text(dx=30, dy=-10, fontSize=14, fontWeight='bold', color='green').encode(
    x='x:O',
    y='y:Q',
    text='text:N'
)

chart = line + text
chart
```

## Year-over-Year Growth Analysis

```python
# Calculate and display YoY growth
data['YoY Growth (%)'] = data['Population (Millions)'].pct_change() * 100

growth_chart = alt.Chart(data.dropna()).mark_bar().encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('YoY Growth (%):Q', title='Year-over-Year Growth (%)'),
    color=alt.condition(
        alt.datum['YoY Growth (%)'] > 0,
        alt.value('steelblue'),
        alt.value('red')
    ),
    tooltip=['Year', alt.Tooltip('YoY Growth (%)', format='.2f')]
).properties(
    title='Year-over-Year Population Growth',
    width=400,
    height=200
)

growth_chart
```
